# Threading

```python
import threading
import datetime
```

```python
def f(i):
    print(f'Thread with id = {i} is working.')

for i in range(0, 3):
    threading.Thread(target=f, args=(i,)).start()


t = threading.Thread(target=f, args=(1,))
t.start()
```

```python
def g(i, name='default'):
    print(f'Thread named {name} with id = {i} is working.')

for i in range(0, 3):
    threading.Thread(target=g, args=(i,), kwargs={'name': f'{i}{i}{i}'}).start()
```

```python
import time
from threading import Thread

def f():
    print(f"The thread \"{threading.current_thread().name}\" started.")
    time.sleep(1)
    print(f"The thread \"{threading.current_thread().name}\" finished.")

t = Thread(name='hello', target=f)

t.daemon = True

t.start()

time.sleep(0.1)
for i in range(5):
    print(i)
```

```python
import threading
import time

def work():
    time.sleep(2)
    print("worker done")

t = threading.Thread(target=work)
t.start()

print("waiting...")
t.join()
print("main continues")
```
